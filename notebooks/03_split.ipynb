# 03 â€” Stratified Train/Val/Test Split

Split `metadata_clean.csv` (9 classes) ke **70 / 15 / 15** dengan `StratifiedShuffleSplit` sehingga distribusi kelas identik di ketiga split.

Output:
- `artifacts/splits/train.csv`
- `artifacts/splits/val.csv`
- `artifacts/splits/test.csv`

Masing-masing berisi kolom `[gambar, laporan, label, label_id, row_id]` dimana `row_id` adalah index stabil untuk alignment dengan cache embedding.

```python
import sys
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedShuffleSplit

sys.path.insert(0, '../src')
from crm import TARGET_CLASSES, LABEL2ID

BASE     = Path('../artifacts/crm_jakarta')
SPLITDIR = Path('../artifacts/splits')
SPLITDIR.mkdir(parents=True, exist_ok=True)

SEED = 42
TRAIN_FRAC, VAL_FRAC, TEST_FRAC = 0.70, 0.15, 0.15
```

## 1. Load cleaned metadata

```python
df = pd.read_csv(BASE / 'metadata_clean.csv')
df['label_id'] = df['label'].map(LABEL2ID)
df['row_id']   = np.arange(len(df))

assert df['label_id'].notna().all(), 'Unmapped label(s) detected'
print(f'Total: {len(df):,} rows')
print(df['label'].value_counts().reindex(TARGET_CLASSES).to_string())
```

## 2. Stratified split

```python
# Step 1: split off test (15%)
sss1 = StratifiedShuffleSplit(n_splits=1, test_size=TEST_FRAC, random_state=SEED)
trainval_idx, test_idx = next(sss1.split(df, df['label_id']))

# Step 2: split remaining 85% â†’ train (70/85) + val (15/85)
trainval_df = df.iloc[trainval_idx].reset_index(drop=True)
sss2 = StratifiedShuffleSplit(n_splits=1, test_size=VAL_FRAC / (TRAIN_FRAC + VAL_FRAC), random_state=SEED)
train_idx, val_idx = next(sss2.split(trainval_df, trainval_df['label_id']))

train_df = trainval_df.iloc[train_idx].reset_index(drop=True)
val_df   = trainval_df.iloc[val_idx].reset_index(drop=True)
test_df  = df.iloc[test_idx].reset_index(drop=True)

print(f'Train: {len(train_df):,} ({len(train_df)/len(df):.1%})')
print(f'Val  : {len(val_df):,} ({len(val_df)/len(df):.1%})')
print(f'Test : {len(test_df):,} ({len(test_df)/len(df):.1%})')
```

## 3. Verify stratification

```python
def dist(d):
    return d['label'].value_counts(normalize=True).reindex(TARGET_CLASSES).round(4)

comp = pd.DataFrame({
    'full':  dist(df),
    'train': dist(train_df),
    'val':   dist(val_df),
    'test':  dist(test_df),
})
print('Class proportion per split (should be nearly identical):')
print(comp.to_string())

max_diff = (comp.max(axis=1) - comp.min(axis=1)).max()
print(f'\nMax cross-split proportion diff: {max_diff:.4f} (target < 0.005)')
```

## 4. Save splits

```python
cols = ['row_id', 'gambar', 'laporan', 'label', 'label_id']
train_df[cols].to_csv(SPLITDIR / 'train.csv', index=False)
val_df[cols].to_csv(SPLITDIR / 'val.csv', index=False)
test_df[cols].to_csv(SPLITDIR / 'test.csv', index=False)
print(f'Saved to {SPLITDIR}/')
for f in sorted(SPLITDIR.glob('*.csv')):
    print(f'  {f.name}: {sum(1 for _ in open(f)) - 1:,} rows')
```
