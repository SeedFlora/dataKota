# 05 — Extract Text Embeddings (GPU)

Run on **RunPod 4090**. Encoders:
- `mE5_large` (paper best)
- `indobert`
- `cendol_mt5`
- `bge_m3` (new SOTA multilingual)
- `nusabert_large` (new Indonesian-optimized)

Runtime estimate (80k laporan, max_length=256, RTX 4090 fp16): ~5-10 menit per encoder.

mE5 pakai instruction prefix otomatis via `TextEncoderSpec.prefix`.

```python
import sys
from pathlib import Path
from time import time

import numpy as np
import pandas as pd
import torch
from tqdm.auto import tqdm

sys.path.insert(0, '../src')
from crm.encoders.text import TEXT_ENCODERS, load_encoder, encode_batch

BASE    = Path('../artifacts/crm_jakarta')
OUT_DIR = Path('../artifacts/embeddings/text')
OUT_DIR.mkdir(parents=True, exist_ok=True)

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Device: {DEVICE}')
```

## 1. Load metadata

```python
df = pd.read_csv(BASE / 'metadata_clean.csv').reset_index(drop=True)
df['laporan'] = df['laporan'].astype(str).str.strip()
print(f'Total rows: {len(df):,}')

# Light preprocessing: strip URLs, excessive whitespace
import re
url_re = re.compile(r'https?://\S+|www\.\S+')
ws_re  = re.compile(r'\s+')
df['laporan_clean'] = df['laporan'].apply(
    lambda t: ws_re.sub(' ', url_re.sub('', t)).strip()
)
print('Sample cleaned laporan:')
print(df['laporan_clean'].head(2).to_list())
```

## 2. Extraction loop

```python
def extract(encoder_name: str, batch_size: int = 32):
    spec = TEXT_ENCODERS[encoder_name]
    out_path = OUT_DIR / f'{encoder_name}.npy'
    if out_path.exists():
        arr = np.load(out_path)
        if arr.shape[0] == len(df):
            print(f'[skip] {encoder_name} already exists: {out_path} shape={arr.shape}')
            return

    print(f'\n=== {encoder_name} | {spec.hf_id} | dim={spec.dim} | pool={spec.pool} ===')
    t0 = time()
    model, tokenizer = load_encoder(spec, DEVICE)
    print(f'Model loaded in {time()-t0:.1f}s')

    embeddings = np.zeros((len(df), spec.dim), dtype=np.float32)
    t0 = time()

    texts_all = df['laporan_clean'].tolist()
    for start in tqdm(range(0, len(texts_all), batch_size), desc=encoder_name):
        end = min(start + batch_size, len(texts_all))
        batch = texts_all[start:end]
        feats = encode_batch(model, tokenizer, batch, spec, DEVICE, use_amp=True)
        embeddings[start:end] = feats

    np.save(out_path, embeddings)
    print(f'Saved {out_path} | shape={embeddings.shape} | time={(time()-t0)/60:.1f} min')

    del model
    if DEVICE == 'cuda':
        torch.cuda.empty_cache()
```

## 3. Run each encoder

```python
extract('mE5_large', batch_size=32)
```

```python
extract('bge_m3', batch_size=32)
```

```python
extract('indobert', batch_size=32)
```

```python
extract('nusabert_large', batch_size=32)
```

```python
# Cendol mT5 — encoder-only, bigger model, smaller batch
extract('cendol_mt5', batch_size=16)
```

## 4. Verify outputs

```python
for f in sorted(OUT_DIR.glob('*.npy')):
    arr = np.load(f, mmap_mode='r')
    norms = np.linalg.norm(arr[:100], axis=1)
    print(f'{f.name:25s} shape={arr.shape} dtype={arr.dtype} L2-norm~{norms.mean():.3f}')
```
