## 0. Setup (same as notebook 07)

```python
import os
os.environ['HF_HOME'] = '/workspace/.cache/huggingface'
os.environ['HF_HUB_ENABLE_HF_TRANSFER'] = '1'

import sys
from pathlib import Path
from time import time

import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
from tqdm.auto import tqdm
from torch.utils.data import Dataset, DataLoader
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

sys.path.insert(0, '../src')
from crm import TARGET_CLASSES
from crm.fusion import early_fusion, class_weights
from crm.encoders.image import IMAGE_ENCODERS

BASE     = Path('../artifacts/crm_jakarta')
IMG_EMB  = Path('../artifacts/embeddings/image')
TXT_EMB  = Path('../artifacts/embeddings/text')
SPLITDIR = Path('../artifacts/splits')
MODELDIR = Path('../artifacts/models')
FIGDIR   = Path('../artifacts/figures')
CHECKPOINT_DIR = MODELDIR / 'checkpoints'

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Device: {DEVICE}')
```

## 1. Extract DINOv3 embeddings

Kalau error `403 gated repo` â†’ access belum approved, balik ke HF request form.

```python
df = pd.read_csv(BASE / 'metadata_clean.csv').reset_index(drop=True)
df['row_id'] = np.arange(len(df))

train = pd.read_csv(SPLITDIR / 'train.csv')
val   = pd.read_csv(SPLITDIR / 'val.csv')
test  = pd.read_csv(SPLITDIR / 'test.csv')

class _ImgDS(Dataset):
    def __init__(self, df, base, preprocess):
        self.paths = [base / p for p in df['gambar'].tolist()]
        self.preprocess = preprocess
    def __len__(self): return len(self.paths)
    def __getitem__(self, idx):
        try:
            img = Image.open(self.paths[idx]).convert('RGB')
            return idx, self.preprocess(img), None
        except Exception as e:
            return idx, None, str(e)

def _collate(batch):
    valid = [(i, t) for i, t, e in batch if t is not None]
    if not valid: return [], None
    idxs, tensors = zip(*valid)
    return list(idxs), torch.stack(list(tensors))

def _pool(out):
    if hasattr(out, 'pooler_output') and out.pooler_output is not None:
        return out.pooler_output
    if hasattr(out, 'last_hidden_state'):
        return out.last_hidden_state[:, 0]
    return out

def extract_dinov3(batch_size=256, num_workers=4):
    out_path = IMG_EMB / 'dinov3_large.npy'
    if out_path.exists() and np.load(out_path, mmap_mode='r').shape[0] == len(df):
        print('[skip] dinov3_large cached')
        return
    spec = IMAGE_ENCODERS['dinov3_large']
    print(f'Loading dinov3_large from HF (gated â€” requires access)...')
    model, preprocess = spec.loader(DEVICE); model.eval()
    ds = _ImgDS(df, BASE, preprocess)
    loader = DataLoader(ds, batch_size=batch_size, num_workers=num_workers,
                        collate_fn=_collate, prefetch_factor=4,
                        persistent_workers=True, pin_memory=True)
    embeddings = np.zeros((len(df), spec.dim), dtype=np.float32)
    with torch.inference_mode():
        for idxs, tensors in tqdm(loader, desc='dinov3_large'):
            if tensors is None: continue
            tensors = tensors.to(DEVICE, non_blocking=True)
            with torch.autocast(device_type=DEVICE.split(':')[0], dtype=torch.float16,
                                enabled=DEVICE.startswith('cuda')):
                feats = _pool(model(tensors))
            embeddings[idxs] = feats.float().cpu().numpy()
    np.save(out_path, embeddings)
    print(f'Saved {out_path} shape={embeddings.shape}')
    del model
    if DEVICE == 'cuda': torch.cuda.empty_cache()

extract_dinov3(batch_size=256, num_workers=8)
```

## 2. Load embeddings + training helpers

```python
img_embs = {f.stem: np.load(f, mmap_mode='r') for f in sorted(IMG_EMB.glob('*.npy'))}
txt_embs = {f.stem: np.load(f, mmap_mode='r') for f in sorted(TXT_EMB.glob('*.npy'))}
print('Image encoders:', list(img_embs.keys()))
print('Text encoders: ', list(txt_embs.keys()))
assert 'dinov3_large' in img_embs, 'DINOv3 embedding belum ada â€” run cell sebelumnya'

def build_features(img_name, txt_name, split):
    idx = split['row_id'].values
    parts = []
    if img_name is not None: parts.append(img_embs[img_name][idx])
    if txt_name is not None: parts.append(txt_embs[txt_name][idx])
    if len(parts) == 2:
        return early_fusion(parts[0], parts[1], l2_per_modality=True)
    return parts[0].astype(np.float32)

def pgs_predict_proba(model, X_test, n_virtual_ensembles=30):
    preds = model.virtual_ensembles_predict(
        X_test, prediction_type='VirtEnsembles',
        virtual_ensembles_count=n_virtual_ensembles)
    preds = np.asarray(preds)
    preds_exp = np.exp(preds - preds.max(axis=-1, keepdims=True))
    probs_per_ens = preds_exp / preds_exp.sum(axis=-1, keepdims=True)
    probs = probs_per_ens.mean(axis=1)
    uncertainty = probs_per_ens.std(axis=1).mean(axis=1)
    return probs, uncertainty

def _fit_catboost(X_tr, y_tr, X_va, y_va, w_list, iterations, depth, posterior_sampling):
    params = dict(depth=depth, learning_rate=0.05, task_type='CPU', thread_count=-1,
                  class_weights=w_list, loss_function='MultiClass',
                  verbose=False, random_seed=42, posterior_sampling=posterior_sampling)
    params['iterations'] = iterations * 2 if posterior_sampling else iterations
    model = CatBoostClassifier(**params)
    t0 = time()
    es = 200 if posterior_sampling else 50
    model.fit(X_tr, y_tr, eval_set=(X_va, y_va), early_stopping_rounds=es)
    return model, time() - t0

def train_and_eval(img_name, txt_name, iterations=1500, depth=6,
                   n_virtual_ensembles=30, save_dir=None, return_models=False):
    X_tr, y_tr = build_features(img_name, txt_name, train), train['label_id'].values
    X_va, y_va = build_features(img_name, txt_name, val),   val['label_id'].values
    X_te, y_te = build_features(img_name, txt_name, test),  test['label_id'].values
    weights = class_weights(y_tr, num_classes=len(TARGET_CLASSES))
    w_list = [weights[i] for i in range(len(TARGET_CLASSES))]
    tag = f'{img_name}__{txt_name}'

    model_cb, t_cb = _fit_catboost(X_tr, y_tr, X_va, y_va, w_list, iterations, depth, False)
    y_pred_cb = model_cb.predict(X_te).flatten().astype(int)
    acc_cb = accuracy_score(y_te, y_pred_cb)
    f1_cb  = f1_score(y_te, y_pred_cb, average='macro')

    model_pgs, t_pgs = _fit_catboost(X_tr, y_tr, X_va, y_va, w_list, iterations, depth, True)
    probs_pgs, uncertainty = pgs_predict_proba(model_pgs, X_te, n_virtual_ensembles)
    y_pred_pgs = probs_pgs.argmax(axis=1)
    acc_pgs = accuracy_score(y_te, y_pred_pgs)
    f1_pgs  = f1_score(y_te, y_pred_pgs, average='macro')

    if save_dir is not None:
        save_dir = Path(save_dir); save_dir.mkdir(parents=True, exist_ok=True)
        model_cb.save_model(save_dir / f'{tag}__cb.cbm')
        model_pgs.save_model(save_dir / f'{tag}__pgs.cbm')

    result = {
        'image': img_name, 'text': txt_name,
        'acc_catboost': acc_cb, 'f1_catboost': f1_cb,
        'acc_pgs': acc_pgs, 'f1_pgs': f1_pgs,
        'delta_acc': acc_pgs - acc_cb, 'delta_f1': f1_pgs - f1_cb,
        'mean_uncertainty': float(np.mean(uncertainty)),
        'train_sec_cb': t_cb, 'train_sec_pgs': t_pgs,
    }
    if return_models:
        result.update({'model_cb': model_cb, 'model_pgs': model_pgs,
                       'probs_pgs': probs_pgs, 'uncertainty': uncertainty,
                       'y_te': y_te, 'y_pred_pgs': y_pred_pgs, 'y_pred_cb': y_pred_cb})
    else:
        del model_cb, model_pgs
    return result
```

## 3. Run DINOv3 Ã— 4 text encoders (incremental)

```python
# backup existing results
existing_path = MODELDIR / 'matrix_results.csv'
if existing_path.exists():
    existing = pd.read_csv(existing_path)
    existing.to_csv(MODELDIR / 'matrix_results.backup.csv', index=False)
    print(f'Backup saved. Existing: {len(existing)} combos')
else:
    existing = pd.DataFrame()
    print('No existing results â€” creating fresh')

new_results = []
for txt in txt_embs.keys():
    img = 'dinov3_large'
    # skip kalau sudah ada
    if len(existing) > 0:
        mask = (existing['image'] == img) & (existing['text'] == txt)
        if mask.any():
            print(f'[skip] {img} + {txt} already done')
            continue
    print(f'\n=== {img} + {txt} ===')
    r = train_and_eval(img, txt, iterations=1500, depth=6, save_dir=CHECKPOINT_DIR)
    new_results.append(r)
    print(f"  CatBoost: acc={r['acc_catboost']:.4f}, f1={r['f1_catboost']:.4f}")
    print(f"  +PGS:     acc={r['acc_pgs']:.4f},     f1={r['f1_pgs']:.4f}   Î”acc={r['delta_acc']:+.4f}")

# merge
if new_results:
    new_df = pd.DataFrame(new_results)
    results_df = pd.concat([existing, new_df], ignore_index=True) if len(existing) > 0 else new_df
else:
    results_df = existing

if 'acc_pgs' in results_df.columns:
    results_df = results_df.sort_values('acc_pgs', ascending=False)
results_df.to_csv(existing_path, index=False)

print('\n=== Updated Top 5 (with DINOv3) ===')
cols = [c for c in ['image','text','acc_catboost','acc_pgs','delta_acc','mean_uncertainty'] if c in results_df.columns]
print(results_df.head(5)[cols].to_string(index=False))
```

## 4. Check â€” apakah DINOv3 jadi winner baru?

```python
best = results_df.iloc[0]
best_img, best_txt = best['image'], best['text']
print(f'Winner: {best_img} + {best_txt} (acc_pgs={best["acc_pgs"]:.4f})')

dinov3_best = results_df[results_df['image'] == 'dinov3_large'].iloc[0] if (results_df['image'] == 'dinov3_large').any() else None
if dinov3_best is not None:
    print(f'DINOv3 best: {dinov3_best["text"]} â†’ acc_pgs={dinov3_best["acc_pgs"]:.4f}')

is_new_winner = best_img == 'dinov3_large'
print(f'\nDINOv3 is new winner? {is_new_winner}')
if is_new_winner:
    print('â†’ Run cell berikutnya untuk refresh ablation + error analysis')
else:
    print('â†’ Winner tetap sama, ablation lama masih valid. Bisa stop di sini.')
```

## 5. (Kondisional) Re-run ablation + error analysis kalau DINOv3 jadi winner

```python
if is_new_winner:
    # Modality ablation
    abl = []
    for label, img, txt in [('image only', best_img, None),
                             ('text only',  None, best_txt),
                             ('fusion',     best_img, best_txt)]:
        print(f'--- {label} ---')
        r = train_and_eval(img, txt, iterations=1500, depth=6)
        r['config'] = label
        abl.append(r)
        print(f"  acc_pgs={r['acc_pgs']:.4f}")
    abl_df = pd.DataFrame(abl)[['config','acc_catboost','f1_catboost','acc_pgs','f1_pgs']]
    abl_df.to_csv(MODELDIR / 'modality_ablation.csv', index=False)
    print('\n=== Modality Ablation ===')
    print(abl_df.to_string(index=False))
else:
    print('Skipped â€” ablation lama masih valid')
```

```python
if is_new_winner:
    # Final + error analysis
    final = train_and_eval(best_img, best_txt, iterations=1500, depth=6,
                           save_dir=CHECKPOINT_DIR, return_models=True)
    final['model_pgs'].save_model(MODELDIR / f'best_{best_img}__{best_txt}.cbm')

    pgs_df = pd.DataFrame([
        {'model': 'CatBoost (no PGS)', 'accuracy': final['acc_catboost'], 'macro_f1': final['f1_catboost']},
        {'model': 'CatBoost + PGS',    'accuracy': final['acc_pgs'],      'macro_f1': final['f1_pgs']},
    ])
    pgs_df.to_csv(MODELDIR / 'pgs_ablation.csv', index=False)
    print('=== PGS Effect ===')
    print(pgs_df.to_string(index=False))

    y_te, y_pred, uncertainty = final['y_te'], final['y_pred_pgs'], final['uncertainty']
    print('\n=== Classification Report ===')
    print(classification_report(y_te, y_pred, target_names=TARGET_CLASSES, digits=4))

    cm = confusion_matrix(y_te, y_pred)
    fig, ax = plt.subplots(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=TARGET_CLASSES, yticklabels=TARGET_CLASSES, ax=ax)
    ax.set_title(f'Confusion Matrix â€” {best_img} + {best_txt} + PGS')
    plt.setp(ax.get_xticklabels(), rotation=30, ha='right')
    plt.tight_layout()
    plt.savefig(FIGDIR / 'confusion_matrix_dinov3.png', dpi=150, bbox_inches='tight')
    plt.show()
else:
    print('Skipped â€” error analysis lama masih valid')
```
