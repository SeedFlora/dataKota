# 04 â€” Extract Image Embeddings (GPU)

Run this on **RunPod 4090**. Untuk setiap image encoder:
1. Load model ke GPU (fp16)
2. Iterate seluruh dataset (train + val + test combined by `row_id`)
3. Save ke `artifacts/embeddings/image/<encoder>.npy`

Order file `.npy` mengikuti `row_id` dari `metadata_clean.csv` â€” aman untuk dipakai lintas split.

**Target encoders:** `dinov2_large`, `hiera_large`, `eva02_large`, `dinov3_large`

Runtime estimate on RTX 4090 (fp16):
- dinov2_large (224): ~15 min
- hiera_large (224): ~12 min
- eva02_large (448): ~30-40 min
- dinov3_large (224): ~15-20 min

```python
import sys
from pathlib import Path
from time import time

import numpy as np
import pandas as pd
import torch
from PIL import Image
from tqdm.auto import tqdm

sys.path.insert(0, '../src')
from crm.encoders.image import IMAGE_ENCODERS, encode_batch

BASE    = Path('../artifacts/crm_jakarta')
IMG_DIR = BASE
OUT_DIR = Path('../artifacts/embeddings/image')
OUT_DIR.mkdir(parents=True, exist_ok=True)

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Device: {DEVICE}')
if DEVICE == 'cuda':
    print(f'GPU: {torch.cuda.get_device_name(0)} | VRAM: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB')
```

## 1. Load metadata (full)

```python
df = pd.read_csv(BASE / 'metadata_clean.csv').reset_index(drop=True)
df['row_id'] = np.arange(len(df))
print(f'Total images to encode: {len(df):,}')
df.head(2)
```

## 2. Extraction loop

Skip jika file `.npy` sudah ada (resumable).

```python
from torch.utils.data import Dataset, DataLoader
from tqdm.auto import tqdm
class _ImgDS(Dataset):
    def __init__(self, df, base, preprocess):
        self.paths = [base / p for p in df['gambar'].tolist()]
        self.preprocess = preprocess
    def __len__(self): return len(self.paths)
    def __getitem__(self, idx):
        try:
            img = Image.open(self.paths[idx]).convert('RGB')
            return idx, self.preprocess(img), None
        except Exception as e:
            return idx, None, str(e)

def _collate(batch):
    valid = [(i, t) for i, t, e in batch if t is not None]
    errs  = [(i, e) for i, t, e in batch if e is not None]
    if not valid:
        return [], None, errs
    idxs, tensors = zip(*valid)
    return list(idxs), torch.stack(list(tensors)), errs

def _pool(out):
    if hasattr(out, 'pooler_output') and out.pooler_output is not None:
        return out.pooler_output
    if hasattr(out, 'last_hidden_state'):
        return out.last_hidden_state[:, 0]
    if isinstance(out, torch.Tensor):
        return out
    raise ValueError(f'Unknown output: {type(out)}')

def extract(encoder_name: str, batch_size: int = 256, num_workers: int = 4, use_amp: bool = True):
    out_path = OUT_DIR / f'{encoder_name}.npy'
    if out_path.exists():
        arr = np.load(out_path)
        if arr.shape[0] == len(df):
            print(f'[skip] {encoder_name} exists: shape={arr.shape}')
            return

    spec = IMAGE_ENCODERS[encoder_name]
    print(f'\n=== {encoder_name} | dim={spec.dim} | bs={batch_size} | workers={num_workers} ===')
    t0 = time()
    model, preprocess = spec.loader(DEVICE)
    model.eval()
    print(f'Model loaded in {time()-t0:.1f}s')

    ds = _ImgDS(df, BASE, preprocess)
    loader = DataLoader(
        ds, batch_size=batch_size, num_workers=num_workers,
        collate_fn=_collate, prefetch_factor=4,
        persistent_workers=True, pin_memory=True,
    )

    embeddings = np.zeros((len(df), spec.dim), dtype=np.float32)
    failures = []
    t0 = time()

    amp_dtype = torch.float16 if use_amp and DEVICE.startswith('cuda') else torch.float32
    dev_type  = DEVICE.split(':')[0]

    with torch.inference_mode():
        for idxs, tensors, errs in tqdm(loader, desc=encoder_name):
            failures.extend(errs)
            if tensors is None:
                continue
            tensors = tensors.to(DEVICE, non_blocking=True)
            with torch.autocast(device_type=dev_type, dtype=amp_dtype,
                                enabled=(amp_dtype != torch.float32)):
                feats = _pool(model(tensors))
            embeddings[idxs] = feats.float().cpu().numpy()

    np.save(out_path, embeddings)
    elapsed = time() - t0
    print(f'Saved {out_path} | shape={embeddings.shape} | time={elapsed/60:.1f} min')
    if failures:
        print(f'Failures: {len(failures)} â€” sample: {failures[:3]}')

    del model
    if DEVICE == 'cuda':
        torch.cuda.empty_cache()
```
