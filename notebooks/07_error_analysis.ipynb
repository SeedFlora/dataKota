# 07 — Error Analysis & Confusion Matrix

Replicate paper's **§IV-C Error Analysis** (Tabel 9) + additions:
- Confusion matrix (full & normalized)
- Per-class precision / recall / F1
- Cases where text-only was wrong but multimodal correct (proves image's contribution)
- Low-confidence samples (PGS uncertainty) — should align with misclassifications

```python
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from catboost import CatBoostClassifier
from sklearn.metrics import confusion_matrix, classification_report
from PIL import Image

sys.path.insert(0, '../src')
from crm import TARGET_CLASSES
from crm.fusion import early_fusion, class_weights
from crm.pgs import pgs_predict_proba

BASE     = Path('../artifacts/crm_jakarta')
IMG_DIR  = Path('../artifacts/embeddings/image')
TXT_DIR  = Path('../artifacts/embeddings/text')
SPLITDIR = Path('../artifacts/splits')
MODELDIR = Path('../artifacts/models')

test = pd.read_csv(SPLITDIR / 'test.csv')
results_df = pd.read_csv(MODELDIR / 'matrix_results.csv').sort_values('acc_pgs', ascending=False)
BEST_IMG, BEST_TXT = results_df.iloc[0][['image', 'text']]
print(f'Best combo: {BEST_IMG} + {BEST_TXT}')
```

## 1. Load best model + predict test set

```python
img_emb = np.load(IMG_DIR / f'{BEST_IMG}.npy', mmap_mode='r')
txt_emb = np.load(TXT_DIR / f'{BEST_TXT}.npy', mmap_mode='r')

idx = test['row_id'].values
X_te = early_fusion(img_emb[idx], txt_emb[idx], l2_per_modality=True)
y_te = test['label_id'].values

model = CatBoostClassifier()
model.load_model(MODELDIR / f'best_catboost_{BEST_IMG}_{BEST_TXT}.cbm')

probs, uncertainty = pgs_predict_proba(model, X_te, n_virtual_ensembles=10)
y_pred = probs.argmax(axis=1)
confidence = probs.max(axis=1)
```

## 2. Confusion matrix

```python
cm = confusion_matrix(y_te, y_pred, labels=list(range(len(TARGET_CLASSES))))
cm_norm = cm / cm.sum(axis=1, keepdims=True).clip(min=1)

short = [lbl[:20] + '…' if len(lbl) > 20 else lbl for lbl in TARGET_CLASSES]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax1,
            xticklabels=short, yticklabels=short, cbar=False)
ax1.set_title('Confusion Matrix (counts)', fontweight='bold')
ax1.set_xlabel('Predicted'); ax1.set_ylabel('True')

sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Blues', ax=ax2,
            xticklabels=short, yticklabels=short, cbar=False, vmin=0, vmax=1)
ax2.set_title('Confusion Matrix (row-normalized)', fontweight='bold')
ax2.set_xlabel('Predicted'); ax2.set_ylabel('True')

for ax in (ax1, ax2):
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
plt.tight_layout()
plt.savefig('confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
print(classification_report(y_te, y_pred, target_names=TARGET_CLASSES, digits=4))
```

## 3. Kasus text-only salah, multimodal benar (Tabel 9 paper)

Train text-only model buat komparasi.

```python
from sklearn.metrics import accuracy_score
import pandas as pd
train = pd.read_csv(SPLITDIR / 'train.csv')
val   = pd.read_csv(SPLITDIR / 'val.csv')

def train_textonly():
    X_tr = txt_emb[train['row_id'].values].astype(np.float32)
    X_va = txt_emb[val['row_id'].values].astype(np.float32)
    y_tr = train['label_id'].values; y_va = val['label_id'].values
    weights = class_weights(y_tr, num_classes=len(TARGET_CLASSES))
    m = CatBoostClassifier(iterations=1000, depth=6, learning_rate=0.05,
                           task_type='GPU', devices='0',
                           class_weights=[weights[i] for i in range(len(TARGET_CLASSES))],
                           posterior_sampling=True, loss_function='MultiClass',
                           verbose=False)
    m.fit(X_tr, y_tr, eval_set=(X_va, y_va), early_stopping_rounds=50)
    return m

text_model = train_textonly()
X_te_text = txt_emb[idx].astype(np.float32)
y_pred_text = text_model.predict(X_te_text).flatten().astype(int)
print(f'Text-only accuracy: {accuracy_score(y_te, y_pred_text):.4f}')
```

```python
# Cases where text was wrong but fusion was correct
wrong_text_right_fusion = (y_pred_text != y_te) & (y_pred == y_te)
print(f'Cases text WRONG, fusion CORRECT: {wrong_text_right_fusion.sum()} / {len(y_te)}')

# Show 6 random examples
idx_cases = np.where(wrong_text_right_fusion)[0]
sample = np.random.RandomState(42).choice(idx_cases, size=min(6, len(idx_cases)), replace=False)

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
for ax, i in zip(axes.flatten(), sample):
    row = test.iloc[i]
    try:
        img = Image.open(BASE / row['gambar']).convert('RGB')
        img.thumbnail((300, 300))
        ax.imshow(img)
    except Exception:
        ax.set_facecolor('#eee')
    true_lbl = TARGET_CLASSES[y_te[i]][:25]
    pred_text = TARGET_CLASSES[y_pred_text[i]][:25]
    ax.set_title(f'Laporan: {row["laporan"][:60]}…\nTrue: {true_lbl}\nText-only: {pred_text} ✗ | Fusion: ✓', fontsize=8)
    ax.axis('off')
plt.suptitle('Text-only wrong → Fusion correct (image contributes)', fontweight='bold')
plt.tight_layout()
plt.savefig('text_wrong_fusion_right.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 4. PGS uncertainty vs correctness

High uncertainty samples should correspond to more misclassifications.

```python
correct = (y_pred == y_te)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist([uncertainty[correct], uncertainty[~correct]], bins=40, label=['Correct', 'Wrong'],
             color=['#3a7bd5', '#d53a3a'], alpha=0.7)
axes[0].set_xlabel('PGS uncertainty (mutual info)')
axes[0].set_ylabel('Count')
axes[0].set_title('Uncertainty distribution')
axes[0].legend()

# Accuracy vs confidence buckets
buckets = np.linspace(0, 1, 11)
bucket_acc = []
for lo, hi in zip(buckets[:-1], buckets[1:]):
    mask = (confidence >= lo) & (confidence < hi)
    bucket_acc.append(correct[mask].mean() if mask.sum() else np.nan)
axes[1].bar(buckets[:-1], bucket_acc, width=0.08, color='#3a7bd5', edgecolor='white')
axes[1].plot([0, 1], [0, 1], 'k--', alpha=0.4, label='perfect calibration')
axes[1].set_xlabel('Predicted confidence')
axes[1].set_ylabel('Actual accuracy')
axes[1].set_title('Calibration curve')
axes[1].legend()
plt.tight_layout()
plt.savefig('uncertainty_analysis.png', dpi=150)
plt.show()
```

## 5. Routing rule berdasarkan confidence

Production rule: kalau `confidence < threshold`, teruskan ke manual review (bukan auto-assign).

```python
thresholds = np.linspace(0.3, 0.95, 14)
rows = []
for t in thresholds:
    auto_mask = confidence >= t
    if auto_mask.sum() == 0: continue
    auto_acc = correct[auto_mask].mean()
    coverage = auto_mask.mean()
    rows.append({'threshold': round(t, 2), 'auto_coverage': coverage,
                 'auto_accuracy': auto_acc, 'manual_needed': 1 - coverage})
thresh_df = pd.DataFrame(rows)
print(thresh_df.to_string(index=False))

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(thresh_df['threshold'], thresh_df['auto_accuracy'], '-o', label='Auto-assignment accuracy', color='#3a7bd5')
ax.plot(thresh_df['threshold'], thresh_df['auto_coverage'], '-s', label='Coverage (auto-rate)', color='#d53a3a')
ax.set_xlabel('Confidence threshold')
ax.set_ylabel('Rate')
ax.set_title('Precision-vs-Coverage trade-off')
ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig('threshold_tradeoff.png', dpi=150)
plt.show()
```
