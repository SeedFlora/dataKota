# 02 â€” Metadata Preprocessing (9-class, match paper)

**Goal:** reduce raw `metadata.csv` â†’ three columns (`gambar`, `laporan`, `label`) dengan **9 target class** sesuai paper:

8 instansi top + `Instansi lain` sebagai catchall untuk laporan yang tidak match top-8 (dipakai model untuk belajar 'bukan top-8' â†’ robust di inference).

**Output:** `artifacts/crm_jakarta/metadata_clean.csv`

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

sns.set_theme(style='whitegrid', context='notebook')
plt.rcParams['figure.dpi'] = 110

BASE = Path('../artifacts/crm_jakarta')
RAW_CSV   = BASE / 'metadata.csv'
CLEAN_CSV = BASE / 'metadata_clean.csv'
```

## 1. Load raw metadata

```python
raw = pd.read_csv(RAW_CSV, low_memory=False)
print(f'Raw rows      : {len(raw):,}')
print(f'Unique SKPD   : {raw["label_skpd"].nunique()}')
```

## 2. Normalize SKPD â†’ 9 target classes

**Mapping rules (ordered â€” first match wins):**

| Target class | Pattern |
|---|---|
| Dinas Perhubungan | `PERHUBUNGAN` OR `PERPARKIRAN` OR `PENGENDALIAN LALU LINTAS` (UP Perparkiran & UPSL masuk sini) |
| Dinas Bina Marga | `BINA MARGA` |
| Satuan Polisi Pamong Praja | `POLISI PAMONG PRAJA` / `SATPOL` |
| Kelurahan | `KELURAHAN` |
| Dinas Pertamanan dan Hutan | `PERTAMANAN` / `HUTAN` |
| Dinas Sumber Daya Air | `SUMBER DAYA AIR` |
| Dinas Cipta Karya, Tata Ruang, dan Pertanahan | `CIPTA KARYA` / `TATA RUANG` / `PERTANAHAN` |
| Badan Pembinaan BUMD | `BUMD` / `BADAN USAHA MILIK DAERAH` / `PT â€¦` / `PERUMDA â€¦` (TransJakarta, Pasar Jaya, PAM Jaya dll masuk sini) |
| Instansi lain | sisanya |

```python
TARGET_CLASSES = [
    'Dinas Bina Marga',
    'Satuan Polisi Pamong Praja',
    'Dinas Perhubungan',
    'Kelurahan',
    'Dinas Pertamanan dan Hutan',
    'Dinas Sumber Daya Air',
    'Dinas Cipta Karya, Tata Ruang, dan Pertanahan',
    'Badan Pembinaan Badan Usaha Milik Daerah',
    'Instansi lain',
]

def normalize_skpd(s: str) -> str:
    s = str(s).upper()
    # Dishub family: dinas + UP Perparkiran + UPSL
    if any(k in s for k in ('PERHUBUNGAN', 'PERPARKIRAN', 'PENGENDALIAN LALU LINTAS', 'LALU LINTAS JALAN')):
        return 'Dinas Perhubungan'
    if 'BINA MARGA' in s:
        return 'Dinas Bina Marga'
    if 'POLISI PAMONG PRAJA' in s or 'SATPOL' in s:
        return 'Satuan Polisi Pamong Praja'
    if 'KELURAHAN' in s:
        return 'Kelurahan'
    if 'PERTAMANAN' in s or 'HUTAN' in s:
        return 'Dinas Pertamanan dan Hutan'
    if 'SUMBER DAYA AIR' in s:
        return 'Dinas Sumber Daya Air'
    if 'CIPTA KARYA' in s or 'TATA RUANG' in s or 'PERTANAHAN' in s:
        return 'Dinas Cipta Karya, Tata Ruang, dan Pertanahan'
    # BUMD: governance body + PT Jakarta enterprises + Perumda
    if ('BUMD' in s or 'BADAN USAHA MILIK DAERAH' in s
            or s.startswith('PT ')
            or s.startswith('PERUMDA')):
        return 'Badan Pembinaan Badan Usaha Milik Daerah'
    return 'Instansi lain'

raw['label'] = raw['label_skpd'].apply(normalize_skpd)
```

## 3. Reduce to 3 columns + save

```python
clean = raw[['gambar', 'laporan', 'label']].copy()
clean = clean.dropna(subset=['gambar', 'laporan', 'label'])
clean['laporan'] = clean['laporan'].astype(str).str.strip()
clean = clean[clean['laporan'].str.len() > 0].reset_index(drop=True)

before = len(clean)
# Exact dedup on laporan text â€” removes spam/copy-paste reports
clean = clean.drop_duplicates(subset='laporan').reset_index(drop=True)
print(f'Exact dedup: {before:,} â†’ {len(clean):,} (dropped {before - len(clean):,} duplicates)')

clean.to_csv(CLEAN_CSV, index=False)
print(f'Saved to : {CLEAN_CSV}')
clean.head(3)
```

## 4. Tabel contoh data (Tabel 1 style)

```python
from matplotlib.offsetbox import OffsetImage, AnnotationBbox
import textwrap

def show_sample_table(df, n=3, seed=0):
    sample = df.sample(n, random_state=seed).reset_index(drop=True)
    fig, ax = plt.subplots(figsize=(11, 1.7 * n + 0.8))
    ax.axis('off')
    headers = ['Gambar', 'Laporan', 'Label']
    xs = [0.08, 0.42, 0.82]
    y_top = 0.92
    for x, h in zip(xs, headers):
        ax.text(x, y_top, h, fontweight='bold', ha='center', transform=ax.transAxes)
    ax.plot([0, 1], [y_top - 0.06, y_top - 0.06], color='black', lw=0.7, transform=ax.transAxes, clip_on=False)
    step = 0.85 / n
    for i, row in sample.iterrows():
        y = y_top - 0.15 - i * step
        try:
            img = Image.open(BASE / row['gambar']).convert('RGB')
            img.thumbnail((140, 140))
            ab = AnnotationBbox(OffsetImage(img, zoom=0.75), (xs[0], y),
                                xycoords='axes fraction', frameon=True, pad=0.1)
            ax.add_artist(ab)
        except Exception:
            ax.text(xs[0], y, '[no img]', ha='center', transform=ax.transAxes)
        wrapped = '\n'.join(textwrap.wrap(row['laporan'][:220], width=45))
        ax.text(xs[1] - 0.15, y, wrapped, ha='left', va='center', fontsize=9, transform=ax.transAxes)
        ax.text(xs[2], y, row['label'], ha='center', va='center', fontsize=9, fontweight='bold', transform=ax.transAxes)
    plt.suptitle('Tabel 1: Contoh Data', y=0.03, fontsize=10)
    plt.tight_layout()
    plt.savefig('sample_table.png', dpi=180, bbox_inches='tight')
    plt.show()

show_sample_table(clean[clean['label'] != 'Instansi lain'], n=3, seed=7)
```

## 5. Distribusi Target Class (Tabel 2 style)

```python
counts = clean['label'].value_counts().reindex(TARGET_CLASSES)
summary = counts.reset_index()
summary.columns = ['Instansi', 'Jumlah']
summary['%'] = (summary['Jumlah'] / summary['Jumlah'].sum() * 100).round(2)
print(summary.to_string(index=False))

pct_top8 = summary.loc[summary['Instansi'] != 'Instansi lain', '%'].sum()
print(f'\nTop-8 coverage: {pct_top8:.2f}%')
```

```python
fig, ax = plt.subplots(figsize=(10, 5.5))
sorted_counts = counts.sort_values()
palette = sns.color_palette('crest', n_colors=8)
color_map = {lbl: palette[i] for i, lbl in enumerate(TARGET_CLASSES[:8])}
color_map['Instansi lain'] = '#b0b0b0'
colors = [color_map[l] for l in sorted_counts.index]
bars = ax.barh(sorted_counts.index, sorted_counts.values, color=colors, edgecolor='white')
for bar, val in zip(bars, sorted_counts.values):
    ax.text(val + sorted_counts.max() * 0.01, bar.get_y() + bar.get_height()/2,
            f'{val:,}', va='center', fontsize=9)
ax.set_xlabel('Jumlah laporan')
ax.set_title('Distribusi 9 Target Class (abu = Instansi lain / catchall)', pad=12, fontweight='bold')
ax.set_xlim(0, sorted_counts.max() * 1.12)
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.savefig('target_class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 6. Class imbalance

```python
maj, minr = counts.max(), counts.min()
print(f'Largest class : {counts.idxmax()} â†’ {maj:,}')
print(f'Smallest class: {counts.idxmin()} â†’ {minr:,}')
print(f'Imbalance ratio (max/min): {maj / minr:.1f}Ã—')
print()
print('Paper pakai class weight: w_l = N / (C * N_l)')
N, C = counts.sum(), len(counts)
weights = {lbl: N / (C * counts[lbl]) for lbl in counts.index}
for lbl, w in sorted(weights.items(), key=lambda x: -x[1]):
    print(f'  {lbl:50s} â†’ w = {w:.3f}')
```

## 7. Panjang laporan (untuk tuning max_length tokenizer)

```python
clean['text_len'] = clean['laporan'].str.len()
# token estimate: ~4 chars per token for Indonesian
clean['est_tokens'] = (clean['text_len'] / 4).round().astype(int)
print('Panjang laporan (karakter):')
print(clean['text_len'].describe(percentiles=[.5, .9, .95, .98, .99]).round(0).to_string())
print()
print('Estimasi token (chars/4):')
print(clean['est_tokens'].describe(percentiles=[.5, .9, .95, .98, .99]).round(0).to_string())
print()
for t in [64, 128, 256, 512]:
    cov = (clean['est_tokens'] <= t).mean() * 100
    print(f'  max_length={t:3d} tokens  â†’  cover {cov:.2f}% laporan')
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
order = counts.index.tolist()
sns.boxplot(data=clean, x='text_len', y='label', order=order, orient='h',
            color='#3a7bd5', fliersize=1, ax=ax)
ax.set_xlim(0, clean['text_len'].quantile(0.98))
ax.set_xlabel('Panjang laporan (karakter)')
ax.set_ylabel('')
ax.set_title('Panjang Laporan per Instansi', fontweight='bold')
sns.despine()
plt.tight_layout()
plt.savefig('text_length_by_class.png', dpi=150)
plt.show()
```

## 8. Contoh gambar per target class (9 classes)

```python
fig, axes = plt.subplots(3, 3, figsize=(12, 12))
axes = axes.flatten()
for i, label in enumerate(TARGET_CLASSES):
    subset = clean[clean['label'] == label]
    row = subset.sample(1, random_state=i).iloc[0]
    try:
        img = Image.open(BASE / row['gambar']).convert('RGB')
        img.thumbnail((320, 320))
        axes[i].imshow(img)
    except Exception:
        axes[i].set_facecolor('#eee')
    short = label if len(label) < 32 else label[:30] + 'â€¦'
    axes[i].set_title(f'{short}\n(n={len(subset):,})', fontsize=9)
    axes[i].axis('off')
plt.suptitle('Contoh gambar per target class (9)', fontsize=12, y=1.00, fontweight='bold')
plt.tight_layout()
plt.savefig('sample_per_class.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 9. Ringkasan

| | Rows | Labels |
|---|---|---|
| Raw `metadata.csv` | 88,302 | 487 SKPD |
| Post-dedup | ~61,773 | 9 |
| **`metadata_clean.csv`** | ~61,773 | **9** (8 top + Instansi lain) |

**Key decisions:**
- **Exact dedup** pada kolom `laporan` â€” hilangkan spam/copy-paste (30% dataset!)
  - E.g. KELURAHAN SEMANAN: 10,884 â†’ 368 rows (satu laporan di-copy 2,208Ã—)
- 9 class training match paper â€” `Instansi lain` jadi catchall
- BUMD digabung dengan PT Jakarta entities + Perumda
- UP Perparkiran & UPSL Lalu Lintas di-bucket ke Dinas Perhubungan
- Class weight: `w_l = N / (C Ã— N_l)` (paper formula)
- Metric: accuracy + macro-F1 (lebih sound untuk imbalanced)
